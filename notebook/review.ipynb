```python
def goles(equipo):
    equipo = pd.Series([equipo]).str.capitalize()[0]
    total_goles = datos.loc[datos['Team'] == equipo, 'Goals'].sum()
    return total_goles    
```

```python
def jugadores_info(nombre):
    """
    Devuelve la información de los jugadores.
    """
    datos['Player'] = datos['Player'].str.lower()  # Normalizar nombres de jugadores a minúsculas
    nombre = nombre.lower()  # Normalizar el nombre del jugador a minúsculas
    df_player = datos[datos['Player'] == nombre]
    return df_player.to_dict(orient='records')
```

```python
def goleadores(opcion):
    """
    Devuelve los goleadores de la liga.
    """
    if opcion == 'top_10':
        gol_10= datos['Goals'].iloc[10]
        return datos[datos['Goals'] >= gol_10]
    elif opcion == 'top_5':
        gol_5= datos['Goals'].iloc[5]
        return datos[datos['Goals'] >= gol_5]
    elif opcion == 'top_3':
        gol_3= datos['Goals'].iloc[3]
        return datos[datos['Goals'] >= gol_3]
    elif opcion == 'top_1':
        gol_1= datos['Goals'].iloc[1]
        return datos[datos['Goals'] >= gol_1]
    else:
        return "Opción no válida. Por favor, elige entre 'top_10', 'top_5', 'top_3' o 'top_1'."

goleadores('top_10')  # Ejemplo de uso
 
```
